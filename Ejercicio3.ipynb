10. Revisión de valores atípicos

```
variables_numericas = ["Weight", "Cholesterol"]

for columna in variables_numericas:
    media = df[columna].mean()
    desviacion = df[columna].std()

    limite_inferior = media - 2 * desviacion
    limite_superior = media + 2 * desviacion

    atipicos = df[
        (df[columna] < limite_inferior) |
        (df[columna] > limite_superior)
    ]

    print("Variable:", columna)
    print("Media:", round(media, 2))
    print("Desviación estándar:", round(desviacion, 2))
    print("Límite inferior:", round(limite_inferior, 2))
    print("Límite superior:", round(limite_superior, 2))
    print("Cantidad de valores atípicos:", len(atipicos))
    print("-" * 40)
```

10.1 Interpretación para Weight
Para la variable Weight, serían considerados atípicos los pesos menores a 81.90 libras o mayores a 205.25 libras. En el archivo analizado, el peso mínimo es 95 y el máximo es 203, por lo tanto no se identifican valores atípicos.


10.2 Interpretación para Cholesterol
Para la variable Cholesterol, serían considerados atípicos los valores menores a 92.14 o mayores a 248.73. En el archivo analizado, el valor mínimo es 102 y el máximo es 235, por lo tanto no se identifican valores atípicos.
